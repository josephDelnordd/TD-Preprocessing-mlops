## Exercice 5 — Imputation pour traiter les données manquantes

```python
origin_Data = pd.read_csv(
    "../data/melb_data.csv"
)
```

```python
origin_Data.info()
```

```python
origin_Data.shape
```

```python
origin_Data.isnull().sum()
```

```python
origin_Data.isnull().sum(axis=1)
```

```python
def missing_columns(originDB):

    colsWithMissing = [
        col
        for col in originDB.columns
        if originDB[col].isnull().any()
    ]

    reduced_original_data = originDB.drop(
        colsWithMissing,
        axis=1
    )

    return colsWithMissing, reduced_original_data
```

```python
colsWithMissing, reducedData = (
    missing_columns(origin_Data)
)

colsWithMissing
```

```python
origin_Data.isnull().mean()*100
```

```python
origin_Data.isnull().mean(axis=1)*100
```

```python
new_Data = origin_Data[
    origin_Data.isnull().mean(axis=1)*100 <= 5
]
```

```python
print(
    new_Data
)
```

## Exercice 6 - Imputation avec SimpleImputer

```python
origin_Data_num = (
    origin_Data.select_dtypes(
        include=['number']
    )
)
```

```python
imputer = SimpleImputer(
    strategy='mean'
)
```

```python
imputer.fit(origin_Data_num)
```

```python
imputer.statistics_
```

```python
transform_data = (
    imputer.transform(
        origin_Data_num
    )
)
```

```python
df_transform = pd.DataFrame(
    transform_data,
    columns=origin_Data_num.columns,
    index=origin_Data_num.index
)
```

```python
df_transform.head()
```

## Exercice 7 - Nettoyage du dataset Flicker

```python
flicker = pd.read_csv(
    "../data/flicker.csv"
)
```

```python
flicker.info()
```

```python
flicker.shape
```

```python
flicker.head()
```

```python
flicker.isnull().sum()
```

```python
flicker.duplicated().sum()
```

```python
flicker.dropna(
    how="all",
    inplace=True
)
```

```python
num_cols = flicker.select_dtypes(
    include="number"
)

num_cols = num_cols.dropna(
    axis=1,
    how="all"
)

imputer = SimpleImputer(
    strategy="mean"
)

result = imputer.fit_transform(
    num_cols
)

num_cols_clean = pd.DataFrame(
    result,
    columns=num_cols.columns,
    index=num_cols.index
)

num_cols_clean.head()
```
